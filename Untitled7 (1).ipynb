```python
import pandas as pd
import requests
from requests.adapters import HTTPAdapter, Retry
import xml.etree.ElementTree as ET
from datetime import datetime
import time
import json
from pathlib import Path
from typing import List, Dict, Set, Optional, Tuple
from collections import defaultdict, deque
import logging
from concurrent.futures import ThreadPoolExecutor, as_completed
import threading
import gzip
import pickle
from functools import lru_cache
import queue
import re
import os
import random

# Simple logging to console only
logging.basicConfig(level=logging.INFO, format='%(message)s')
logger = logging.getLogger(__name__)

class TokenBucket:
    """Token bucket rate limiter - O(1) operations without holding locks during sleep"""
    def __init__(self, rate: float, capacity: float = None):
        self.rate = rate  # tokens per second
        self.capacity = capacity or rate
        self.tokens = self.capacity
        self.last_update = time.time()
        self.lock = threading.Lock()

    def consume(self, tokens: int = 1) -> float:
        """
        Consume tokens from bucket. Returns wait time if not enough tokens.
        """
        while True:
            with self.lock:
                now = time.time()
                elapsed = now - self.last_update
                self.tokens = min(self.capacity, self.tokens + elapsed * self.rate)
                self.last_update = now

                if self.tokens >= tokens:
                    self.tokens -= tokens
                    return 0  # No wait needed
                else:
                    # Calculate wait time needed
                    deficit = tokens - self.tokens
                    wait_time = deficit / self.rate
                    return wait_time

class ClinVarParallelAPI:
    def clear_cache(self, gene_names: List[str] = None):
        """
        Clear cache files to force fresh data retrieval
        """
        print("\n[CACHE] Clearing cache files...")

        # Clear variant ID cache for specific genes
        if gene_names:
            cache_key = '_'.join(sorted(gene_names))
            variants_cache = self.cache_dir / f"{cache_key}_variant_ids.json"
            if variants_cache.exists():
                variants_cache.unlink()
                print(f"[CACHE] Deleted variant ID cache for {', '.join(gene_names)}")

        # Clear all XML and parsed caches
        xml_files = list(self.xml_cache_dir.glob("*.xml.gz"))
        pkl_files = list(self.parsed_cache_dir.glob("*.pkl"))

        for f in xml_files[:100]:  # Limit to avoid too much deletion
            f.unlink()
        for f in pkl_files[:100]:
            f.unlink()

        print(f"[CACHE] Deleted {min(100, len(xml_files))} XML cache files")
        print(f"[CACHE] Deleted {min(100, len(pkl_files))} parsed cache files")

    def __init__(self, email: str, api_key: str = None, cache_dir: str = "clinvar_cache_parallel"):
        self.email = email
        self.api_key = api_key or os.environ.get('NCBI_API_KEY')
        self.cache_dir = Path(cache_dir)
        self.cache_dir.mkdir(exist_ok=True)

        # Create subdirectories for different cache types
        self.xml_cache_dir = self.cache_dir / "xml"
        self.xml_cache_dir.mkdir(exist_ok=True)
        self.parsed_cache_dir = self.cache_dir / "parsed"
        self.parsed_cache_dir.mkdir(exist_ok=True)
        self.version_cache_dir = self.cache_dir / "versions"  # Separate cache for versions
        self.version_cache_dir.mkdir(exist_ok=True)
        self.stats_cache = self.cache_dir / "stats.json"

        self.eutils_base = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils"

        # Token bucket rate limiting - increase rate for better throughput
        self.requests_per_second = 10 if self.api_key else 3
        self.rate_limiter = TokenBucket(self.requests_per_second)

        # Set up session with increased connection pool
        self.session = requests.Session()
        self.session.headers.update({"User-Agent": f"ClinVarParallelAPI/1.0 ({self.email})"})

        # Increase connection pool size for better concurrency
        adapter = HTTPAdapter(
            pool_connections=50,
            pool_maxsize=50,
            max_retries=Retry(
                total=8,
                backoff_factor=0.5,
                status_forcelist=[429, 500, 502, 503, 504],
                allowed_methods=["GET"]
            )
        )
        self.session.mount("https://", adapter)
        self.session.mount("http://", adapter)

        # Batch fetching configuration
        self.variant_batch_size = 200
        self.version_batch_size = 200

        # Temporal split configuration
        self.training_cutoff_date = None  # Will be set if temporal split is enabled

        # Debug tracking - EXPANDED
        self.debug_stats = defaultdict(int)

        # RCV level analysis flag
        self.use_rcv_level = False

        # Validation strictness
        self.strict_condition_matching = True
        self.require_pmid_validation = True

        # DEBUG FLAGS
        self.debug_rcv = True  # Enable detailed RCV debugging
        self.debug_scv = True  # Enable detailed SCV debugging
        self.debug_conditions = True  # Enable condition matching debugging

        print(f"[INIT] Parallel ClinVar API initialized")
        print(f"[INIT] Email: {self.email}")
        print(f"[INIT] API Key: {'Configured' if self.api_key else 'Not provided'}")
        print(f"[INIT] Rate limit: {self.requests_per_second} requests/second")
        print(f"[INIT] DEBUG MODE: RCV={self.debug_rcv}, SCV={self.debug_scv}, Conditions={self.debug_conditions}")
        logger.info("ClinVar API initialized successfully")

    def _get(self, url, params):
        """
        Rate-limited HTTP GET with proper token bucket usage
        """
        while True:
            wait_time = self.rate_limiter.consume()
            if wait_time <= 0:
                break
            time.sleep(wait_time)

        # Add API key and email to params
        if self.api_key:
            params.setdefault('api_key', self.api_key)
        params.setdefault('email', self.email)

        # Make request with exponential backoff for 429s
        max_retries = 5
        for attempt in range(max_retries):
            try:
                r = self.session.get(url, params=params, timeout=60)

                if r.status_code == 429:
                    retry_after = r.headers.get("Retry-After")
                    wait = float(retry_after) if retry_after else (2 ** attempt)
                    print(f"[RATE LIMIT] Got 429, waiting {wait} seconds (attempt {attempt+1}/{max_retries})...")
                    time.sleep(wait + random.uniform(0, 0.5))
                    continue

                r.raise_for_status()
                return r

            except requests.exceptions.RequestException as e:
                if attempt == max_retries - 1:
                    logger.error(f"Failed after {max_retries} attempts: {e}")
                    raise
                time.sleep(2 ** attempt)

    def parse_date(self, date_str: str) -> Optional[datetime]:
        """
        Parse various date formats from ClinVar
        """
        if not date_str:
            return None

        # Try different formats
        for fmt in ["%Y-%m-%d", "%Y-%m", "%Y", "%d-%b-%Y", "%B %d, %Y"]:
            try:
                return datetime.strptime(date_str, fmt)
            except:
                continue

        # Try ISO format with time
        try:
            return datetime.fromisoformat(date_str.replace('Z', '+00:00'))
        except:
            pass

        return None

    def search_variants(self, gene_names: List[str], limit: int = None, unbiased: bool = False) -> List[str]:
        """
        Search for variants in specified genes
        """
        print("\n" + "="*80)
        print(f"SEARCHING FOR VARIANTS IN GENES: {', '.join(gene_names)}")
        print("="*80)

        variation_ids = []

        try:
            search_url = f"{self.eutils_base}/esearch.fcgi"

            # Build search query
            if unbiased:
                # Get ALL variants for unbiased analysis
                if len(gene_names) == 1:
                    search_query = f'{gene_names[0]}[gene]'
                else:
                    gene_query = ' OR '.join([f'{gene}[gene]' for gene in gene_names])
                    search_query = f'({gene_query})'
                print(f"[SEARCH] Unbiased search - getting ALL variants")
            else:
                # Original biased search for conflicts
                if len(gene_names) == 1:
                    search_query = f'{gene_names[0]}[gene] AND (conflicting interpretations[Properties] OR multiple submitters[Properties])'
                else:
                    gene_query = ' OR '.join([f'{gene}[gene]' for gene in gene_names])
                    search_query = f'({gene_query}) AND (conflicting interpretations[Properties] OR multiple submitters[Properties])'
                print(f"[SEARCH] Filtered search - variants with conflicts or multiple submitters")

            print(f"[SEARCH] Query: {search_query}")

            # Get count first
            count_params = {
                'db': 'clinvar',
                'term': search_query,
                'retmode': 'json',
                'retmax': 0
            }

            response = self._get(search_url, count_params)
            data = response.json()

            total_count = int(data.get('esearchresult', {}).get('count', 0))
            print(f"[SEARCH] Found {total_count} records in ClinVar")

            if limit:
                retmax = min(limit, total_count)
            else:
                retmax = total_count

            batch_size = 500
            for start in range(0, retmax, batch_size):
                params = {
                    'db': 'clinvar',
                    'term': search_query,
                    'retmode': 'json',
                    'retstart': start,
                    'retmax': min(batch_size, retmax - start)
                }

                response = self._get(search_url, params)
                data = response.json()

                ids = data.get('esearchresult', {}).get('idlist', [])
                variation_ids.extend(ids)

                print(f"[SEARCH] Retrieved {len(variation_ids)}/{retmax} IDs")

                if limit and len(variation_ids) >= limit:
                    break

            return variation_ids[:limit] if limit else variation_ids

        except Exception as e:
            print(f"[ERROR] Search failed: {e}")
            logger.error(f"Search failed: {e}")
            return []

    def fetch_variants_parallel(self, variation_ids: List[str], max_workers: int = 20) -> Dict[str, str]:
        """
        NEW: Parallel variant fetching with caching
        """
        results = {}
        to_fetch = []

        # Check cache first
        for var_id in variation_ids:
            cache_file = self.xml_cache_dir / f"var_{var_id}.xml.gz"
            if cache_file.exists():
                try:
                    with gzip.open(cache_file, 'rt', encoding='utf-8') as f:
                        results[var_id] = f.read()
                        self.debug_stats['variants_from_cache'] += 1
                except:
                    to_fetch.append(var_id)
            else:
                to_fetch.append(var_id)

        if not to_fetch:
            print(f"[BATCH FETCH] All {len(variation_ids)} variants from cache!")
            return results

        print(f"\n[BATCH FETCH] Fetching {len(to_fetch)} variants ({len(results)} from cache) with {max_workers} workers")

        def fetch_single_variant(var_id):
            """Fetch a single variant with caching"""
            try:
                fetch_url = f"{self.eutils_base}/efetch.fcgi"
                params = {
                    'db': 'clinvar',
                    'rettype': 'vcv',
                    'retmode': 'xml',
                    'id': str(var_id).strip(),
                    'from_esearch': 'true',
                    'is_variationid': 'true'
                }

                response = self._get(fetch_url, params)

                if response.status_code == 200 and len(response.text) > 200:
                    if '<VariationArchive' in response.text:
                        # Cache the result
                        cache_file = self.xml_cache_dir / f"var_{var_id}.xml.gz"
                        try:
                            with gzip.open(cache_file, 'wt', encoding='utf-8') as f:
                                f.write(response.text)
                        except:
                            pass

                        self.debug_stats['variants_with_valid_xml'] += 1
                        return var_id, response.text
                    else:
                        self.debug_stats['variants_with_invalid_xml'] += 1
                else:
                    self.debug_stats['variants_fetch_failed'] += 1
            except Exception as e:
                self.debug_stats['variants_fetch_failed'] += 1
                logger.debug(f"Failed to fetch variant {var_id}: {e}")
            return var_id, None

        # Parallel fetching
        with ThreadPoolExecutor(max_workers=max_workers) as executor:
            futures = {executor.submit(fetch_single_variant, vid): vid for vid in to_fetch}

            completed = 0
            for future in as_completed(futures):
                var_id = futures[future]
                try:
                    vid, xml = future.result(timeout=60)
                    if xml:
                        results[vid] = xml
                    completed += 1

                    if completed % 100 == 0 or completed == len(to_fetch):
                        print(f"[BATCH FETCH] Progress: {completed}/{len(to_fetch)} ({100*completed/len(to_fetch):.1f}%)")
                except:
                    pass

        self.debug_stats['variants_fetched_successfully'] = len(results)
        print(f"[BATCH FETCH] Successfully retrieved {len(results)}/{len(variation_ids)} variants")
        print(f"[DEBUG] Valid XML: {self.debug_stats['variants_with_valid_xml']}, Invalid XML: {self.debug_stats['variants_with_invalid_xml']}")
        print(f"[DEBUG] Variants from cache: {self.debug_stats.get('variants_from_cache', 0)}")
        return results

    def fetch_variants_batch(self, variation_ids: List[str], batch_size: int = None) -> Dict[str, str]:
        """
        REPLACED WITH PARALLEL VERSION - Keeping for compatibility
        """
        return self.fetch_variants_parallel(variation_ids, max_workers=20)

    def fetch_versions_parallel(self, version_ids: List[str], max_workers: int = 20) -> Dict[str, str]:
        """
        NEW: Parallel version fetching with caching
        """
        results = {}
        to_fetch = []

        # Check cache first
        for ver_id in version_ids:
            cache_file = self.version_cache_dir / f"{ver_id.replace('.', '_')}.xml.gz"
            if cache_file.exists():
                try:
                    with gzip.open(cache_file, 'rt', encoding='utf-8') as f:
                        results[ver_id] = f.read()
                        self.debug_stats['versions_from_cache'] += 1
                except:
                    to_fetch.append(ver_id)
            else:
                to_fetch.append(ver_id)

        if not to_fetch:
            print(f"[VERSION FETCH] All {len(version_ids)} versions from cache!")
            return results

        print(f"[VERSION FETCH] Fetching {len(to_fetch)} versions ({len(results)} from cache) with {max_workers} workers")

        def fetch_single_version(ver_id):
            """Fetch a single version with caching"""
            try:
                fetch_url = f"{self.eutils_base}/efetch.fcgi"
                params = {
                    'db': 'clinvar',
                    'rettype': 'vcv',
                    'retmode': 'xml',
                    'id': ver_id
                }

                response = self._get(fetch_url, params)

                if response.status_code == 200 and len(response.text) > 200:
                    # Cache the result
                    cache_file = self.version_cache_dir / f"{ver_id.replace('.', '_')}.xml.gz"
                    try:
                        with gzip.open(cache_file, 'wt', encoding='utf-8') as f:
                            f.write(response.text)
                    except:
                        pass

                    self.debug_stats['versions_fetched'] += 1
                    return ver_id, response.text
            except Exception as e:
                logger.debug(f"Failed to fetch version {ver_id}: {e}")
            return ver_id, None

        # Parallel fetching
        with ThreadPoolExecutor(max_workers=max_workers) as executor:
            futures = {executor.submit(fetch_single_version, vid): vid for vid in to_fetch}

            completed = 0
            for future in as_completed(futures):
                ver_id = futures[future]
                try:
                    vid, xml = future.result(timeout=60)
                    if xml:
                        results[vid] = xml
                    completed += 1

                    if completed % 100 == 0 or completed == len(to_fetch):
                        print(f"[VERSION FETCH] Progress: {completed}/{len(to_fetch)} ({100*completed/len(to_fetch):.1f}%)")
                except:
                    pass

        print(f"[VERSION FETCH] Successfully fetched {len(results) - (len(version_ids) - len(to_fetch))}/{len(to_fetch)} new versions")
        return results

    def fetch_versions_batch(self, version_ids: List[str], batch_size: int = None) -> Dict[str, str]:
        """
        REPLACED WITH PARALLEL VERSION - Keeping for compatibility
        """
        return self.fetch_versions_parallel(version_ids, max_workers=20)

    def _first_text(self, root, paths):
        """Helper to find first non-empty text from multiple paths"""
        for p in paths:
            el = root.find(p)
            if el is not None:
                txt = (el.text or '').strip()
                if txt:
                    return txt
        return None

    def extract_variant_type(self, root, variation_id: str = "unknown") -> str:
        """
        Extract variant type - MolecularConsequence first, then HGVS, then structural
        FIXED: No longer maps splice_region_variant to splice_site_variant
        """
        variant_type = 'unknown'

        try:
            # FIRST: Check MolecularConsequence (most precise)
            # FIXED: Removed splice_region_variant
            consequence_priority = {
                'splice_acceptor_variant': 1,
                'splice_donor_variant': 1,
                'stop_gained': 3,
                'nonsense_variant': 3,
                'frameshift_variant': 4,
                'missense_variant': 5,
            }

            best_consequence = None
            best_priority = 999

            for mol_cons in root.findall('.//MolecularConsequence'):
                cons_type = mol_cons.get('Type', '')
                if cons_type:
                    normalized = cons_type.replace(' ', '_').lower()

                    # Track if we're rejecting splice_region
                    if normalized == 'splice_region_variant':
                        self.debug_stats['splice_region_rejected'] += 1
                        continue  # Skip splice_region_variant entirely

                    if normalized in consequence_priority:
                        priority = consequence_priority[normalized]
                        if priority < best_priority:
                            best_consequence = normalized
                            best_priority = priority

            if best_consequence:
                variant_type = best_consequence
                self.debug_stats['variant_types_found'] += 1
                return variant_type

            # SECOND: Parse HGVS expressions
            for hgvs_elem in root.findall('.//HGVS'):
                if hgvs_elem.text and hgvs_elem.text.strip():
                    hgvs_text = hgvs_elem.text.strip()

                    # Protein changes
                    if 'p.' in hgvs_text:
                        if '*' in hgvs_text or 'Ter' in hgvs_text:
                            variant_type = 'nonsense_variant'
                            self.debug_stats['variant_types_found'] += 1
                            return variant_type
                        elif 'fs' in hgvs_text:
                            variant_type = 'frameshift_variant'
                            self.debug_stats['variant_types_found'] += 1
                            return variant_type
                        elif '=' not in hgvs_text and (
                            re.search(r'p\.[A-Z][a-z]{2}\d+[A-Z][a-z]{2}', hgvs_text) or  # 3-letter
                            re.search(r'p\.[A-Z]\d+[A-Z]', hgvs_text)  # 1-letter
                        ):
                            variant_type = 'missense_variant'
                            self.debug_stats['variant_types_found'] += 1
                            return variant_type

                    # DNA changes - Only canonical splice sites ±1-2
                    elif 'c.' in hgvs_text:
                        if re.search(r'c\.\d+[+-][1-2]\b', hgvs_text):  # ±1-2 only
                            variant_type = 'splice_site_variant'
                            self.debug_stats['variant_types_found'] += 1
                            return variant_type

            # THIRD: Check for large deletions
            # FIXED: Require explicit size for all deletions
            var_archive = root.find('.//VariationArchive')
            if var_archive is not None:
                var_type_attr = (var_archive.get('VariationType', '') or '').lower()

                if 'copy number loss' in var_type_attr or 'deletion' in var_type_attr:
                    size = self._get_cnv_size_bp(root)

                    # FIXED: Only accept if size is known and >= 50
                    if size and size >= 50:
                        variant_type = 'large_deletion'
                        self.debug_stats['variant_types_found'] += 1
                        return variant_type
                    elif size is None:
                        # Track rejected deletions without size
                        self.debug_stats['deletions_rejected_no_size'] += 1

        except Exception as e:
            logger.error(f"Error extracting variant type: {e}")

        # Unknown type (will be filtered out)
        self.debug_stats['variant_types_found'] += 1
        return variant_type

    def is_clinically_meaningful_variant(self, variant_type: str) -> bool:
        if not variant_type:
            return False

        t = variant_type.lower().replace(' ', '_').replace('-', '_')

        # STRICT: Only missense + canonical LoF
        allowed = {
            'missense_variant',
            'nonsense_variant',
            'stop_gained',
            'frameshift_variant',
            'splice_acceptor_variant',
            'splice_donor_variant',
            'splice_site_variant',  # Only canonical ±1-2 now
            'large_deletion'  # Only with known size >= 50
        }

        is_meaningful = t in allowed

        if not is_meaningful:
            self.debug_stats['filter_reasons'] += 1
        else:
            self.debug_stats['variants_with_meaningful_type_count'] += 1

        return is_meaningful

    def pre_filter_variants(self, variant_xmls: Dict[str, str], gene_names: List[str]) -> Dict[str, Dict]:
        """
        NEW OPTIMIZATION: Pre-filter variants before fetching versions
        Returns dict of var_id -> basic_info for variants that pass filters
        """
        print(f"\n[PRE-FILTER] Pre-filtering {len(variant_xmls)} variants for efficiency...")

        filtered_variants = {}

        for var_id, xml_content in variant_xmls.items():
            try:
                root = ET.fromstring(xml_content)

                # Check for VariationArchive
                var_archive = root.find('.//VariationArchive')
                if var_archive is None:
                    self.debug_stats['filtered_no_variation_archive'] += 1
                    continue

                # Extract variant type EARLY
                variant_type = self.extract_variant_type(root, var_id)

                # EARLY EXIT if not meaningful
                if not self.is_clinically_meaningful_variant(variant_type):
                    self.debug_stats['filtered_not_meaningful'] += 1
                    self.debug_stats['variants_pre_filtered'] += 1
                    continue

                # Gene verification - more flexible
                gene_symbols = set()
                for gene_elem in root.findall('.//Gene'):
                    # Try both attribute and text
                    symbol = gene_elem.get('Symbol') or gene_elem.findtext('Symbol') or ''
                    if symbol:
                        gene_symbols.add(symbol)

                # Skip if no overlap with requested genes
                if gene_names and not gene_symbols.intersection(set(gene_names)):
                    self.debug_stats['filtered_gene_mismatch'] += 1
                    self.debug_stats['variants_pre_filtered'] += 1
                    continue

                # Store basic info for this variant
                vcv_accession = var_archive.get('Accession', '')
                current_version = int(var_archive.get('Version', '1'))

                filtered_variants[var_id] = {
                    'xml': xml_content,
                    'vcv_accession': vcv_accession,
                    'current_version': current_version,
                    'variant_type': variant_type,
                    'gene_symbols': gene_symbols
                }

            except Exception as e:
                logger.debug(f"Error pre-filtering variant {var_id}: {e}")
                self.debug_stats['variants_pre_filtered'] += 1
                continue

        print(f"[PRE-FILTER] Kept {len(filtered_variants)}/{len(variant_xmls)} variants after pre-filtering")
        print(f"[PRE-FILTER] Filtered out {self.debug_stats['variants_pre_filtered']} variants")

        return filtered_variants

    def parse_clinical_assertion(self, assertion_elem) -> Dict:
        """
        Parse ClinicalAssertion with complete date handling and condition extraction
        FIXED: Use MedGen ID as fallback when condition names are missing
        """
        submission = {}

        try:
            # SCV Accession - keep both full and root
            scv_elem = assertion_elem.find('.//ClinVarAccession[@Type="SCV"]')
            if scv_elem is not None:
                accession = scv_elem.get('Accession', '')
                version = scv_elem.get('Version', '')
                submission['scv_accession'] = f"{accession}.{version}" if accession and version else ""
                submission['scv_root'] = accession  # Root without version for grouping
                submission['submitter'] = scv_elem.get('SubmitterName', '') or scv_elem.get('OrgAbbreviation', '')

                # Get ALL dates with proper fallback
                submission['date_updated'] = scv_elem.get('DateUpdated', '')
                submission['date_created'] = scv_elem.get('DateCreated', '')

                # Parse all dates
                for date_field in ['date_updated', 'date_created']:
                    if submission.get(date_field):
                        parsed_date = self.parse_date(submission[date_field])
                        if parsed_date:
                            submission[f'{date_field}_parsed'] = parsed_date

            # Extract condition/phenotype for proper grouping (with MedGen IDs for better matching)
            condition_parts = []
            condition_ids = []

            # DEBUG: Log what we're looking for
            if self.debug_conditions:
                traits = assertion_elem.findall('.//TraitSet/Trait')
                if len(traits) == 0:
                    print(f"[DEBUG-CONDITION] No Trait elements found in assertion")

            # Try multiple locations for trait information
            for trait in assertion_elem.findall('.//TraitSet/Trait'):
                # Get preferred name - try multiple paths
                name_found = False
                for name_path in ['./Name/ElementValue[@Type="Preferred"]', './Name/ElementValue', './Name']:
                    name_elem = trait.find(name_path)
                    if name_elem is not None:
                        name_text = name_elem.text if name_elem.text else name_elem.get('Value', '')
                        if name_text:
                            condition_parts.append(name_text.strip())
                            name_found = True
                            break

                # Get MedGen ID if available
                xref = trait.find('./XRef[@DB="MedGen"]')
                if xref is not None:
                    medgen_id = xref.get('ID', '')
                    if medgen_id:
                        condition_ids.append(f"MedGen:{medgen_id}")

            # FIXED: Use MedGen ID as condition key if names are missing
            if condition_ids and not condition_parts:
                # Use MedGen ID as the condition key
                submission['condition_key'] = '|'.join(sorted(condition_ids))[:200]
                if self.debug_conditions:
                    print(f"[DEBUG-CONDITION] Using MedGen IDs as key: {submission['condition_key']}")
            elif condition_parts:
                submission['condition_key'] = '|'.join(sorted(condition_parts))[:200]
            else:
                submission['condition_key'] = 'unspecified'
                if self.debug_conditions:
                    print(f"[DEBUG-CONDITION] No conditions found, using 'unspecified'")

            submission['condition_names'] = condition_parts
            submission['condition_ids'] = condition_ids
            submission['condition_id_key'] = '|'.join(sorted(condition_ids)) if condition_ids else ''

            # Classification - handle both old and new XML formats
            classification_found = False

            # Check multiple locations for classification
            class_elem = assertion_elem.find('.//Classification')
            if class_elem is not None:
                submission['date_last_evaluated'] = class_elem.get('DateLastEvaluated', '')

                # Try multiple paths for germline classification
                classification = self._first_text(class_elem, [
                    './GermlineClassification',
                    './GermlineClassification/Description',
                    './/GermlineClassification/Description'
                ])

                if classification:
                    submission['classification'] = classification
                    classification_found = True

            if not classification_found:
                clin_sig_elem = assertion_elem.find('.//ClinicalSignificance')
                if clin_sig_elem is not None:
                    desc_elem = clin_sig_elem.find('./Description')
                    if desc_elem is not None and desc_elem.text:
                        submission['classification'] = desc_elem.text.strip()
                        classification_found = True
                    if not submission.get('date_last_evaluated'):
                        submission['date_last_evaluated'] = clin_sig_elem.get('DateLastEvaluated', '')

            # Parse dates to datetime objects for proper sorting
            if submission.get('date_last_evaluated'):
                parsed_date = self.parse_date(submission['date_last_evaluated'])
                if parsed_date:
                    submission['date_last_evaluated_parsed'] = parsed_date
                    submission['date_last_evaluated'] = parsed_date.strftime('%Y-%m-%d')

            # Citations - FIXED: Deduplicate PMIDs
            pmids_seen = set()
            citations = []
            for citation in assertion_elem.findall('.//Citation'):
                pmid_elem = citation.find('./ID[@Source="PubMed"]')
                if pmid_elem is not None and pmid_elem.text:
                    pmid = pmid_elem.text.strip()
                    if pmid not in pmids_seen:
                        citations.append({'pmid': pmid})
                        pmids_seen.add(pmid)

            submission['citations'] = citations
            submission['citation_count'] = len(citations)

            # Extract method types
            method_types = set()
            for method in assertion_elem.findall('.//ObservedIn/Method/MethodType'):
                if method is not None and method.text:
                    method_types.add(method.text.strip())
            submission['method_types'] = sorted(method_types)

            # Extract functional evidence
            functional_attrs = []
            for attr in assertion_elem.findall('.//Attribute'):
                if attr is not None:
                    attr_type = (attr.get('Type') or '').lower()
                    if attr_type in {'functionalevidence', 'functional_evidence'} and attr.text:
                        functional_attrs.append(attr.text.strip())
            submission['functional_evidence'] = functional_attrs

            if submission.get('scv_accession'):
                return submission

        except Exception as e:
            logger.error(f"Error parsing clinical assertion: {e}")

        return None

    def normalize_classification_category(self, classification: str) -> str:
        """
        FIXED: More granular normalization that preserves Likely vs non-Likely
        """
        s = (classification or '').lower().strip()
        s = s.replace('_', ' ').replace('-', ' ')

        # Handle combined classifications first
        if re.search(r'\bpathogenic\s*/\s*likely\s*pathogenic\b', s):
            return 'pathogenic_combined'

        # Exclude special cases
        if 'conflicting' in s:
            return 'conflicting'
        if 'low penetrance' in s:
            return 'not_relevant'
        if 'risk factor' in s:
            return 'not_relevant'

        # FIXED: Keep likely_pathogenic separate from pathogenic
        if re.search(r'\blikely\b.*\bpathogenic\b', s):
            return 'likely_pathogenic'
        if re.search(r'\bpathogenic\b', s) and 'likely' not in s:
            return 'pathogenic'
        if re.search(r'\blikely\b.*\bbenign\b', s):
            return 'likely_benign'
        if re.search(r'\bbenign\b', s) and 'likely' not in s:
            return 'benign'
        if re.search(r'\buncertain\b', s) or 'vus' in s:
            return 'uncertain_significance'
        if 'not provided' in s:
            return 'not_provided'

        return 'other'

    def conditions_match_strictly(self, cond1_key: str, cond2_key: str,
                                cond1_ids: List[str], cond2_ids: List[str]) -> bool:
        """
        FIXED: Stricter condition matching - reject empty conditions
        """
        # FIXED: If both missing usable IDs and names, do NOT match
        if (not cond1_ids and not cond2_ids) and (not cond1_key and not cond2_key):
            self.debug_stats['empty_condition_mismatches'] += 1
            return False

        # If we have MedGen IDs for both, use those (most accurate)
        if cond1_ids and cond2_ids:
            return set(cond1_ids) == set(cond2_ids)

        # Otherwise, require exact name match
        cond1_norm = (cond1_key or '').lower().strip()
        cond2_norm = (cond2_key or '').lower().strip()

        # Don't match if either is 'unspecified'
        if cond1_norm == 'unspecified' or cond2_norm == 'unspecified':
            return False

        if self.strict_condition_matching:
            # Exact match required
            match = cond1_norm == cond2_norm
            if not match and self.debug_conditions:
                print(f"[DEBUG-CONDITION] Mismatch: '{cond1_norm}' != '{cond2_norm}'")
                self.debug_stats['condition_mismatches'] += 1
            return match
        else:
            # Original lenient matching (not recommended for training)
            if not cond1_norm or not cond2_norm:
                return False  # Changed from True

            # Check word overlap
            words1 = set(cond1_norm.split())
            words2 = set(cond2_norm.split())
            overlap = len(words1 & words2)

            # Require at least 80% overlap
            min_words = min(len(words1), len(words2))
            if min_words > 0:
                return overlap / min_words >= 0.8
            return False

    def classify_change(self, old_class: str, new_class: str, variation_id: str = "unknown") -> str:
        """
        FIXED: More precise classification that distinguishes likely vs non-likely
        """
        old_cat = self.normalize_classification_category(old_class)
        new_cat = self.normalize_classification_category(new_class)

        # VUS to pathogenic spectrum
        if old_cat == 'uncertain_significance' and new_cat == 'likely_pathogenic':
            return 'upgraded_vus_to_likely_pathogenic'
        elif old_cat == 'uncertain_significance' and new_cat == 'pathogenic':
            return 'upgraded_vus_to_pathogenic'
        elif old_cat == 'uncertain_significance' and new_cat == 'pathogenic_combined':
            return 'upgraded_vus_to_pathogenic'

        # Pathogenic spectrum to VUS
        elif old_cat == 'likely_pathogenic' and new_cat == 'uncertain_significance':
            return 'downgraded_likely_pathogenic_to_vus'
        elif old_cat == 'pathogenic' and new_cat == 'uncertain_significance':
            return 'downgraded_pathogenic_to_vus'
        elif old_cat == 'pathogenic_combined' and new_cat == 'uncertain_significance':
            return 'downgraded_pathogenic_to_vus'

        # Within pathogenic spectrum changes
        elif old_cat == 'likely_pathogenic' and new_cat == 'pathogenic':
            return 'upgraded_likely_pathogenic_to_pathogenic'
        elif old_cat == 'pathogenic' and new_cat == 'likely_pathogenic':
            return 'downgraded_pathogenic_to_likely_pathogenic'

        else:
            return 'not_relevant'

    def classify_change_with_condition(self, old_class: str, new_class: str,
                                      old_condition_key: str, new_condition_key: str,
                                      old_condition_ids: List[str], new_condition_ids: List[str],
                                      variation_id: str = "unknown") -> str:
        """
        Classify change only if conditions match (using stricter matching)
        """
        # Check if conditions match
        if not self.conditions_match_strictly(old_condition_key, new_condition_key,
                                             old_condition_ids, new_condition_ids):
            return 'different_condition'

        # Now check classification change
        return self.classify_change(old_class, new_class, variation_id)

    def _pmids_and_study_types_for_condition(self, root, condition_key: str,
                                            target_classification: str,
                                            condition_ids: List[str] = None) -> Tuple[List[str], List[str], Dict[str, Set[str]]]:
        """
        IMPROVED: Returns PMIDs, study types, and mapping of PMID to study types
        FIXED: Deduplicate PMIDs before returning
        """
        pmids = set()
        study_types = set()
        pmid2types = defaultdict(set)

        # Normalize for comparison
        def _norm(s):
            return self.normalize_classification_category(s or '')

        # Parse condition for matching
        cond_tokens = set(condition_key.lower().split('|')) if condition_key else set()

        for assertion in root.findall('.//ClinicalAssertion'):
            # Step 1: Check classification match
            cls = None

            # Try multiple paths for classification (XML 2.0 compatibility)
            class_paths = [
                './/Classification/GermlineClassification',
                './/Classification/GermlineClassification/Description',
                './/ClinicalSignificance/Description'
            ]

            cls = self._first_text(assertion, class_paths)

            if _norm(cls) != _norm(target_classification):
                continue  # Skip if classification doesn't match

            # Step 2: Check condition match (improved)
            assertion_cond_ids = []
            names = []
            for trait in assertion.findall('.//TraitSet/Trait'):
                name_elem = trait.find('./Name/ElementValue[@Type="Preferred"]')
                if name_elem is not None and name_elem.text:
                    names.append(name_elem.text.strip().lower())

                # Get MedGen IDs
                xref = trait.find('./XRef[@DB="MedGen"]')
                if xref is not None:
                    medgen_id = xref.get('ID', '')
                    if medgen_id:
                        assertion_cond_ids.append(f"MedGen:{medgen_id}")

            # Use ID matching if available
            if condition_ids and assertion_cond_ids:
                if not set(condition_ids).intersection(set(assertion_cond_ids)):
                    continue  # Skip if no ID match
            elif cond_tokens and names:
                if not cond_tokens.intersection(set(names)):
                    continue  # Skip if no name match
            elif condition_key and not names:
                continue  # Skip if we're looking for specific condition but none found

            # Step 3: Extract study types for this assertion
            st = set()
            for method in assertion.findall('.//ObservedIn/Method/MethodType'):
                if method is not None and method.text:
                    txt = method.text.strip().lower()
                    if 'functional' in txt or 'in vitro' in txt or 'in vivo' in txt:
                        st.add('functional')
                    elif 'case-control' in txt:
                        st.add('case-control')
                    elif 'segregation' in txt or 'co-segregation' in txt:
                        st.add('segregation')
                    elif 'clinical testing' in txt:
                        st.add('clinical_testing')
                    elif 'literature' in txt:
                        st.add('literature_only')
                    elif 'population' in txt:
                        st.add('population')
                    elif 'computational' in txt or 'in silico' in txt:
                        st.add('computational')

            for attr in assertion.findall('.//Attribute'):
                att = (attr.get('Type') or '').lower()
                if 'functional' in att:
                    st.add('functional')
                if 'computational' in att:
                    st.add('computational')

            # Step 4: Extract PMIDs and map to study types - FIXED: Deduplicate
            for citation in assertion.findall('.//Citation'):
                pmid_elem = citation.find('./ID[@Source="PubMed"]')
                if pmid_elem is not None and pmid_elem.text:
                    pmid = pmid_elem.text.strip()
                    pmids.add(pmid)

                    # Map this PMID to the study types from this assertion
                    types_for_this_pmid = st if st else {'unspecified'}
                    for t in types_for_this_pmid:
                        pmid2types[pmid].add(t)

                    study_types.update(types_for_this_pmid)

                    # Track if we're not validating PMID relevance
                    if not self.require_pmid_validation:
                        self.debug_stats['pmids_without_validation'] += 1

        # Convert to sorted lists and dict
        pmid_to_types_dict = {k: sorted(v) for k, v in pmid2types.items()}
        return sorted(pmids), sorted(study_types), pmid_to_types_dict

    def _extract_rcv_classifications(self, root) -> Dict[str, Dict]:
        """
        Extract RCV-level classifications, selecting latest per condition
        IMPROVED: Better handling of missing dates and XML 2.0
        """
        rcv_data_raw = defaultdict(list)

        # DEBUG: Track RCV elements found
        rcv_elements = root.findall('.//ReferenceClinVarAssertion')
        if self.debug_rcv:
            print(f"[DEBUG-RCV] Found {len(rcv_elements)} ReferenceClinVarAssertion elements")

        # Collect all RCVs
        for rcv in rcv_elements:
            rcv_acc_elem = rcv.find('.//ClinVarAccession[@Type="RCV"]')
            if rcv_acc_elem is None:
                if self.debug_rcv:
                    print(f"[DEBUG-RCV] No RCV ClinVarAccession found in element")
                continue

            rcv_id = rcv_acc_elem.get('Accession', '')
            if self.debug_rcv and rcv_id:
                print(f"[DEBUG-RCV] Processing RCV: {rcv_id}")

            # Extract condition/trait with IDs
            condition_parts = []
            condition_ids = []

            # Try multiple paths for traits
            trait_paths = ['.//TraitSet/Trait', './/Trait']
            traits_found = False

            for trait_path in trait_paths:
                traits = rcv.findall(trait_path)
                if traits:
                    traits_found = True
                    for trait in traits:
                        # Try multiple paths for name
                        name_elem = None
                        for name_path in ['./Name/ElementValue[@Type="Preferred"]', './Name/ElementValue', './Name']:
                            name_elem = trait.find(name_path)
                            if name_elem is not None:
                                name_text = name_elem.text if name_elem.text else name_elem.get('Value', '')
                                if name_text:
                                    condition_parts.append(name_text.strip())
                                    break

                        # Get MedGen ID
                        xref = trait.find('./XRef[@DB="MedGen"]')
                        if xref is not None:
                            medgen_id = xref.get('ID', '')
                            if medgen_id:
                                condition_ids.append(f"MedGen:{medgen_id}")
                    break

            if self.debug_rcv and not traits_found:
                print(f"[DEBUG-RCV] No traits found for RCV {rcv_id}")

            # Use MedGen IDs as key if names are missing
            if condition_ids and not condition_parts:
                condition_key = '|'.join(sorted(condition_ids))[:200]
            elif condition_parts:
                condition_key = '|'.join(sorted(condition_parts))[:200]
            else:
                condition_key = 'unspecified'

            condition_id_key = '|'.join(sorted(condition_ids)) if condition_ids else ''

            # Get classification - try multiple paths
            classification = None
            date_eval = ''

            clin_sig = rcv.find('.//ClinicalSignificance')
            if clin_sig is not None:
                desc_elem = clin_sig.find('./Description')
                if desc_elem is not None and desc_elem.text:
                    classification = desc_elem.text.strip()
                    date_eval = clin_sig.get('DateLastEvaluated', '')

                    if self.debug_rcv:
                        print(f"[DEBUG-RCV] RCV {rcv_id}: classification='{classification}', date='{date_eval}'")

            if classification:
                # FIXED: Use ID-first key
                key_primary = condition_id_key if condition_id_key else condition_key

                # Append to list for this condition
                rcv_data_raw[key_primary].append({
                    'rcv_id': rcv_id,
                    'classification': classification,
                    'date_evaluated': date_eval,
                    'date_parsed': self.parse_date(date_eval) if date_eval else None,
                    'condition_ids': condition_ids,
                    'condition_id_key': condition_id_key,
                    'condition_key': condition_key  # Keep name key for display
                })
            elif self.debug_rcv:
                print(f"[DEBUG-RCV] No classification found for RCV {rcv_id}")

        if self.debug_rcv:
            print(f"[DEBUG-RCV] Collected {len(rcv_data_raw)} unique conditions with RCV data")

        # Select latest per condition (prefer dated over undated)
        rcv_data = {}
        for key_primary, entries in rcv_data_raw.items():
            if entries:
                # Separate dated and undated
                dated = [e for e in entries if e['date_parsed'] is not None]
                undated = [e for e in entries if e['date_parsed'] is None]

                if dated:
                    # Use latest dated entry
                    latest = max(dated, key=lambda x: (x['date_parsed'], x['rcv_id']))
                elif undated:
                    # Fall back to lexicographic RCV ID if no dates
                    latest = max(undated, key=lambda x: x['rcv_id'])
                else:
                    continue

                rcv_data[key_primary] = latest

        if self.debug_rcv:
            print(f"[DEBUG-RCV] Final RCV data contains {len(rcv_data)} conditions")

        return rcv_data

    def build_reclassification_timeline(self, submissions: List[Dict], variation_id: str = "unknown",
                                       use_scv_reclassifications: bool = True) -> Tuple[List[Dict], Dict]:
        """
        IMPROVED: Build timeline with stricter condition matching and per-condition conflicts
        Returns: (reclassifications, current_conflicts)
        """
        def format_date(sub):
            """Get best available date as string"""
            if sub.get('date_last_evaluated'):
                return sub['date_last_evaluated']
            elif sub.get('date_updated'):
                return sub['date_updated']
            elif sub.get('date_created'):
                return sub['date_created']
            return ''

        reclassifications = []
        current_conflicts = {}

        if len(submissions) < 2:
            return reclassifications, current_conflicts

        # Only build SCV reclassifications if flag is True
        if use_scv_reclassifications:
            if self.debug_scv:
                print(f"[DEBUG-SCV] Processing {len(submissions)} submissions for variant {variation_id}")

            # Track changes within same submitter+SCV+condition over time
            by_submission_key = defaultdict(list)
            for sub in submissions:
                submitter = sub.get('submitter', 'Unknown')
                scv_root = sub.get('scv_root', '')
                condition_key = sub.get('condition_key', '')
                condition_id_key = sub.get('condition_id_key', '')

                if submitter and submitter != 'Unknown' and scv_root:
                    # Create composite key including condition IDs
                    key = (submitter, scv_root, condition_key, condition_id_key)

                    # Use all available dates with proper fallback
                    timeline_date = (sub.get('date_last_evaluated_parsed') or
                                   sub.get('date_updated_parsed') or
                                   sub.get('date_created_parsed'))
                    if timeline_date:
                        sub['_timeline_dt'] = timeline_date
                        by_submission_key[key].append(sub)

            # Find TEMPORAL changes within each submission key
            for key, subs in by_submission_key.items():
                if len(subs) < 2:
                    continue

                submitter, scv_root, condition_key, condition_id_key = key

                # Sort by parsed datetime (not string)
                subs.sort(key=lambda x: x.get('_timeline_dt', datetime.min))

                for i in range(1, len(subs)):
                    prev_sub = subs[i-1]
                    curr_sub = subs[i]

                    prev_class = prev_sub.get('classification', '').strip()
                    curr_class = curr_sub.get('classification', '').strip()
                    prev_condition = prev_sub.get('condition_key', '')
                    curr_condition = curr_sub.get('condition_key', '')
                    prev_cond_ids = prev_sub.get('condition_ids', [])
                    curr_cond_ids = curr_sub.get('condition_ids', [])

                    # Skip if no classification data or same classification
                    if not prev_class or not curr_class or prev_class.lower() == curr_class.lower():
                        continue

                    # Check time gap to avoid technical resubmissions
                    time_gap = curr_sub['_timeline_dt'] - prev_sub['_timeline_dt']
                    if time_gap.days < 30:  # Less than 30 days - likely technical update
                        continue

                    # Apply temporal filter if training cutoff is set
                    if self.training_cutoff_date:
                        if curr_sub['_timeline_dt'] > self.training_cutoff_date:
                            self.debug_stats['temporal_filtered'] += 1
                            continue

                    # Use improved condition-aware classification check
                    change_type = self.classify_change_with_condition(
                        prev_class, curr_class,
                        prev_condition, curr_condition,
                        prev_cond_ids, curr_cond_ids,
                        variation_id
                    )

                    # Skip if different conditions or not relevant
                    if change_type in ['different_condition', 'not_relevant']:
                        continue

                    # This is a real temporal reclassification
                    has_citations = len(curr_sub.get('citations', [])) > 0

                    if self.debug_scv:
                        print(f"[DEBUG-SCV] Found reclassification: {prev_class} -> {curr_class} ({change_type})")

                    reclassifications.append({
                        'submitter': submitter,
                        'scv_root': scv_root,
                        'condition': condition_key,
                        'condition_ids': curr_cond_ids,
                        'date_old': format_date(prev_sub),
                        'classification_old': prev_class,
                        'date_new': format_date(curr_sub),
                        'classification_new': curr_class,
                        'scv_old': prev_sub.get('scv_accession', ''),
                        'scv_new': curr_sub.get('scv_accession', ''),
                        'citations_old': prev_sub.get('citations', []),
                        'citations_new': curr_sub.get('citations', []),
                        'has_citations': has_citations,
                        'change_type': change_type,
                        'time_gap_days': time_gap.days,
                        'source': 'SCV',
                        'validation_level': 'strict' if self.strict_condition_matching else 'lenient'
                    })

                    self.debug_stats[f'reclassifications_found_{change_type}'] += 1

        # FIXED: Detect conflicts PER CONDITION
        # Build latest by (submitter, condition)
        latest_by_submitter_and_condition = defaultdict(dict)
        for sub in submissions:
            submitter = sub.get('submitter', 'Unknown')
            ckey = sub.get('condition_key', '')
            cidkey = sub.get('condition_id_key', '')

            if not submitter or submitter == 'Unknown':
                continue

            dt = (sub.get('date_last_evaluated_parsed') or
                  sub.get('date_updated_parsed') or
                  sub.get('date_created_parsed'))

            key = (ckey, cidkey)
            best = latest_by_submitter_and_condition[submitter].get(key)

            if (best is None) or (dt and dt > (best.get('_best_dt') or datetime.min)):
                sub['_best_dt'] = dt
                latest_by_submitter_and_condition[submitter][key] = sub

        # Check conflicts per condition bucket
        per_condition_latest = defaultdict(list)
        for submitter, cond_dict in latest_by_submitter_and_condition.items():
            for cond_key, sub in cond_dict.items():
                per_condition_latest[cond_key].append((submitter, sub))

        # Find conflicts
        for cond_key, items in per_condition_latest.items():
            if len(items) < 2:
                continue

            cats = defaultdict(list)
            for submitter, sub in items:
                cls = self.normalize_classification_category(sub.get('classification', ''))
                if cls and cls != 'not_provided':
                    cats[cls].append((submitter, sub))

            # Check for pathogenic spectrum vs VUS conflicts
            has_path = any(k in ['pathogenic', 'likely_pathogenic', 'pathogenic_combined'] for k in cats)
            has_vus = 'uncertain_significance' in cats

            if has_path and has_vus:
                # Get latest date for this conflict
                all_dates = []
                for _, sub in items:
                    if sub.get('_best_dt'):
                        all_dates.append(sub['_best_dt'])

                latest_dt = max(all_dates) if all_dates else None

                current_conflicts = {
                    'has_vus_path_conflict': True,
                    'conflict_type': 'vus_vs_pathogenic',
                    'condition': cond_key[0] or 'unspecified',
                    'condition_ids': cond_key[1] if len(cond_key) > 1 else '',
                    'classifications': {k: [(s, ss['classification']) for s, ss in v]
                                      for k, v in cats.items()},
                    'num_submitters': sum(len(v) for v in cats.values()),
                    'latest_date': latest_dt.strftime('%Y-%m-%d') if latest_dt else ''
                }
                self.debug_stats['current_conflicts_found'] += 1
                break  # Only report first conflict found

        return reclassifications, current_conflicts

    def _get_cnv_size_bp(self, root) -> Optional[int]:
        """Calculate CNV size in base pairs"""
        locations = root.findall('.//SequenceLocation')
        max_size = None

        for loc in locations:
            try:
                start = int(loc.get('start', '') or loc.get('innerStart', '') or 0)
                stop = int(loc.get('stop', '') or loc.get('innerStop', '') or 0)
                if start and stop and stop >= start:
                    size = stop - start + 1
                    max_size = max(max_size or 0, size)
            except:
                pass

        return max_size

    def extract_version_info(self, root, version_id: str) -> Dict:
        """
        Extract classification and date info from a VCV version
        FIXED: Handle both old and new XML structures
        """
        try:
            # FIXED: Don't use 'or root' fallback
            var_archive = root.find('.//VariationArchive')
            if var_archive is None:
                self.debug_stats['filtered_no_variation_archive'] += 1
                return None

            date_updated = (
                var_archive.get('DateLastUpdated', '') or
                var_archive.get('DateUpdated', '') or
                ''
            )

            # FIXED: Try multiple paths for classification (new XML first)
            classification = self._first_text(var_archive, [
                './Classifications/GermlineClassification/Description',  # XML 2.0
                './Interpretation/ClinicalSignificance/Description',     # Alternative
                './/GermlineClassification/Description',                 # Old fallback
                './ClinicalSignificance/Description'                     # Direct child
            ])

            # Collect ALL submitters (be defensive on structure)
            submitters = set()
            for scv in root.findall('.//ClinicalAssertion//ClinVarAccession[@Type="SCV"]'):
                name = scv.get('SubmitterName') or scv.get('OrgAbbreviation') or ''
                if name:
                    submitters.add(name)

            # Return AFTER collecting all data
            if classification:
                self.debug_stats['versions_with_classification'] += 1
                return {
                    'version_id': version_id,
                    'date': date_updated,
                    'date_parsed': self.parse_date(date_updated),
                    'classification': classification,
                    'submitter_count': len(submitters),
                    'submitters': sorted(list(submitters))[:3]
                }
            else:
                self.debug_stats['versions_without_classification'] += 1

        except Exception as e:
            self.debug_stats['xml_structure_errors'] += 1
            logger.debug(f"Error extracting version info for {version_id}: {e}")

        return None

    def fetch_all_vcv_versions_batch_optimized(self, filtered_variants: Dict[str, Dict]) -> Dict[str, List[Dict]]:
        """
        OPTIMIZED: Only fetch versions for variants with >1 version
        """
        # OPTIMIZATION 1: Filter to multi-version variants only
        multi_version_variants = {}
        single_version_count = 0

        for var_id, var_info in filtered_variants.items():
            if var_info['current_version'] > 1:
                multi_version_variants[var_id] = var_info
            else:
                single_version_count += 1

        print(f"\n[VERSION OPTIMIZATION] {single_version_count} variants have only 1 version (skipping)")
        print(f"[VERSION OPTIMIZATION] {len(multi_version_variants)} variants have multiple versions")

        if not multi_version_variants:
            print("[VERSION OPTIMIZATION] No multi-version variants found - no version history to fetch!")
            return {}

        print(f"\n[VERSION FETCH] Fetching version history for {len(multi_version_variants)} MULTI-VERSION variants only...")

        all_versions_data = {}

        # Initialize cache for version XMLs
        if not hasattr(self, 'version_xmls_cache'):
            self.version_xmls_cache = {}

        # Build list of all version IDs to fetch
        all_version_ids = []
        version_id_to_variant = {}
        expected_versions = {}

        for var_id, var_info in multi_version_variants.items():
            vcv_accession = var_info['vcv_accession']
            current_version = var_info['current_version']

            if vcv_accession:
                # Cache the current version XML
                current_version_id = f"{vcv_accession}.{current_version}"
                self.version_xmls_cache[current_version_id] = var_info['xml']
                expected_versions[var_id] = current_version

                # Generate all version IDs to fetch
                for v in range(1, current_version + 1):
                    version_id = f"{vcv_accession}.{v}"
                    all_version_ids.append(version_id)
                    version_id_to_variant[version_id] = var_id

        # Calculate actual savings
        total_possible_versions = sum(v['current_version'] for v in filtered_variants.values())
        actual_versions_to_fetch = len(all_version_ids)
        versions_saved = total_possible_versions - actual_versions_to_fetch

        print(f"[VERSION OPTIMIZATION] Total possible versions: {total_possible_versions}")
        print(f"[VERSION OPTIMIZATION] Actually fetching: {actual_versions_to_fetch}")
        print(f"[VERSION OPTIMIZATION] Saved {versions_saved} version fetches by skipping single-version variants!")

        # OPTIMIZATION 2: Use parallel fetching with caching
        if all_version_ids:
            print(f"[VERSION FETCH] Using parallel fetching with caching...")
            version_xmls = self.fetch_versions_parallel(all_version_ids, max_workers=20)

            # Parse version data
            for version_id, xml_content in version_xmls.items():
                if xml_content:
                    var_id = version_id_to_variant.get(version_id)
                    if not var_id:
                        continue

                    try:
                        root = ET.fromstring(xml_content)
                        version_info = self.extract_version_info(root, version_id)

                        if version_info:
                            if var_id not in all_versions_data:
                                all_versions_data[var_id] = []
                            all_versions_data[var_id].append(version_info)

                    except Exception as e:
                        logger.error(f"Error parsing version {version_id}: {e}")

        # Sort versions and check completeness
        for var_id in all_versions_data:
            versions = all_versions_data[var_id]

            def get_version_num(v):
                try:
                    return int(v['version_id'].split('.')[-1])
                except:
                    return 0

            versions.sort(
                key=lambda x: (
                    x.get('date_parsed') is None,
                    x.get('date_parsed') or datetime.min,
                    get_version_num(x)
                )
            )

            # Check completeness
            if var_id in expected_versions:
                expected = expected_versions[var_id]
                parsed_versions = [get_version_num(v) for v in versions]
                max_parsed = max(parsed_versions) if parsed_versions else 0

                if max_parsed < expected:
                    missing = list(set(range(1, expected + 1)) - set(parsed_versions))
                    for v in versions:
                        v['version_history_incomplete'] = True
                        v['missing_versions'] = missing

                    self.debug_stats['incomplete_histories'] += 1

        print(f"[VERSION FETCH] Retrieved version history for {len(all_versions_data)} variants")
        print(f"[VERSION FETCH] Versions from cache: {self.debug_stats.get('versions_from_cache', 0)}")
        return all_versions_data

    def process_variants_parallel_optimized(self, variation_ids: List[str], gene_names: List[str],
                                           max_workers: int = 10, fetch_versions: bool = True,
                                           require_citations: bool = False,
                                           use_scv_reclassifications: bool = False) -> List[Dict]:
        """
        OPTIMIZED: Two-pass processing - filter first, then fetch versions
        """
        print(f"\n[PROCESSING] Optimized two-pass processing of {len(variation_ids)} variants")

        results = []
        start_time = time.time()

        # PASS 1: Fetch all variant XMLs IN PARALLEL
        print("\n[PASS 1] Fetching all variant XMLs IN PARALLEL...")
        variant_xmls = self.fetch_variants_parallel(variation_ids, max_workers=20)
        print(f"[PASS 1] Fetched XML for {len(variant_xmls)} variants")

        # PASS 2: Pre-filter to meaningful variants
        print("\n[PASS 2] Pre-filtering variants...")
        filtered_variants = self.pre_filter_variants(variant_xmls, gene_names)
        print(f"[PASS 2] {len(filtered_variants)} variants passed pre-filtering")

        # PASS 3: Fetch version histories ONLY for filtered variants with >1 version
        version_histories = {}
        if fetch_versions and len(filtered_variants) > 0:
            print("\n[PASS 3] Fetching version histories for MULTI-VERSION variants only...")
            version_histories = self.fetch_all_vcv_versions_batch_optimized(filtered_variants)
            print(f"[PASS 3] Retrieved version histories for {len(version_histories)} variants")

        # PASS 4: Process the filtered variants with their version histories
        print("\n[PASS 4] Processing variant data...")

        def process_variant_with_versions(var_id: str, var_info: Dict) -> Optional[Dict]:
            """Process a pre-filtered variant with its version history"""
            try:
                xml_content = var_info['xml']
                root = ET.fromstring(xml_content)
                self.debug_stats['variants_processed'] += 1

                result = {
                    'variation_id': var_id,
                    'name': '',
                    'rsid': '',
                    'variant_type': var_info['variant_type'],  # Already extracted
                    'submissions': [],
                    'reclassifications': [],
                    'current_conflicts': {},
                    'vcv_version_changes': [],
                    'rcv_version_changes': [],
                    'version_history_incomplete': False
                }

                # Get basic variant info
                var_archive = root.find('.//VariationArchive')
                result['vcv_accession'] = var_info['vcv_accession']

                # Get variant name
                var_name = var_archive.get('VariationName', '').strip()
                if var_name:
                    result['name'] = var_name

                # Get rsID
                xref_elem = root.find('.//SimpleAllele/XRefList/XRef[@DB="dbSNP"]')
                if xref_elem is not None:
                    rs_id = xref_elem.get('ID', '')
                    if rs_id:
                        result['rsid'] = f"rs{rs_id}" if not rs_id.startswith('rs') else rs_id

                # Parse clinical assertions
                seen_scvs = set()
                assertions = root.findall('.//ClinicalAssertion')
                for assertion in assertions:
                    submission = self.parse_clinical_assertion(assertion)
                    if submission and submission.get('scv_accession'):
                        if submission['scv_accession'] not in seen_scvs:
                            seen_scvs.add(submission['scv_accession'])
                            result['submissions'].append(submission)

                # Build SCV reclassification timeline
                reclassifications, conflicts = self.build_reclassification_timeline(
                    result['submissions'],
                    variation_id=var_id,
                    use_scv_reclassifications=use_scv_reclassifications
                )

                if require_citations:
                    reclassifications = [r for r in reclassifications if r.get('has_citations')]

                result['reclassifications'] = reclassifications
                result['current_conflicts'] = conflicts

                # Add version changes if available
                if var_id in version_histories and len(version_histories[var_id]) > 1:
                    versions_data = version_histories[var_id]

                    # Check completeness
                    result['version_history_incomplete'] = any(
                        v.get('version_history_incomplete', False) for v in versions_data
                    )

                    if result.get('version_history_incomplete'):
                        self.debug_stats['variants_with_incomplete_history'] += 1

                    # Process version changes
                    def extract_version_number(version_info):
                        try:
                            return int(version_info['version_id'].split('.')[-1])
                        except:
                            return None

                    for i in range(1, len(versions_data)):
                        prev_version = versions_data[i-1]
                        curr_version = versions_data[i]

                        prev_version_num = extract_version_number(prev_version)
                        curr_version_num = extract_version_number(curr_version)

                        # RCV-level analysis (if enabled)
                        if self.use_rcv_level and hasattr(self, 'version_xmls_cache'):
                            prev_xml = self.version_xmls_cache.get(prev_version['version_id'])
                            curr_xml = self.version_xmls_cache.get(curr_version['version_id'])

                            if prev_xml and curr_xml:
                                try:
                                    prev_root = ET.fromstring(prev_xml)
                                    curr_root = ET.fromstring(curr_xml)

                                    # Get RCV-level classifications
                                    prev_rcv_map = self._extract_rcv_classifications(prev_root)
                                    curr_rcv_map = self._extract_rcv_classifications(curr_root)

                                    if self.debug_rcv and (prev_rcv_map or curr_rcv_map):
                                        print(f"[DEBUG-RCV] Version {prev_version_num}: {len(prev_rcv_map)} RCVs")
                                        print(f"[DEBUG-RCV] Version {curr_version_num}: {len(curr_rcv_map)} RCVs")

                                    # Compare per condition
                                    for condition_key in set(prev_rcv_map.keys()) | set(curr_rcv_map.keys()):
                                        prev_rcv = prev_rcv_map.get(condition_key, {})
                                        curr_rcv = curr_rcv_map.get(condition_key, {})

                                        if not prev_rcv or not curr_rcv:
                                            continue

                                        # Apply temporal filter
                                        curr_eval_dt = curr_rcv.get('date_parsed')
                                        if self.training_cutoff_date and curr_eval_dt and curr_eval_dt > self.training_cutoff_date:
                                            self.debug_stats['temporal_filtered'] += 1
                                            continue

                                        prev_class = prev_rcv.get('classification', '').strip()
                                        curr_class = curr_rcv.get('classification', '').strip()

                                        if prev_class and curr_class and prev_class.lower() != curr_class.lower():
                                            change_type = self.classify_change(prev_class, curr_class, var_id)

                                            if change_type and change_type != 'not_relevant':
                                                # Get PMIDs with study type mapping
                                                condition_ids = curr_rcv.get('condition_ids', [])
                                                prev_pmids, _, prev_map = self._pmids_and_study_types_for_condition(
                                                    prev_root, condition_key, prev_class, condition_ids
                                                )
                                                curr_pmids, _, curr_map = self._pmids_and_study_types_for_condition(
                                                    curr_root, condition_key, curr_class, condition_ids
                                                )

                                                # Calculate new PMIDs and their study types
                                                new_linked_pmids = sorted(set(curr_pmids) - set(prev_pmids))
                                                study_types_new_linked = sorted({t for p in new_linked_pmids
                                                                                for t in curr_map.get(p, [])} or ['unspecified'])

                                                if self.debug_rcv:
                                                    print(f"[DEBUG-RCV] Found RCV change: {prev_class} -> {curr_class}")

                                                rcv_change = {
                                                    'version_old': prev_version_num,
                                                    'version_new': curr_version_num,
                                                    'condition': curr_rcv.get('condition_key', condition_key),
                                                    'condition_ids': condition_ids,
                                                    'rcv_old': prev_rcv.get('rcv_id', ''),
                                                    'rcv_new': curr_rcv.get('rcv_id', ''),
                                                    'date_old': prev_rcv.get('date_evaluated', ''),
                                                    'classification_old': prev_class,
                                                    'date_new': curr_rcv.get('date_evaluated', ''),
                                                    'classification_new': curr_class,
                                                    'change_type': change_type,
                                                    'source': 'RCV',
                                                    'linked_pmids': curr_pmids,
                                                    'new_linked_pmids': new_linked_pmids,
                                                    'study_types_linked_new': study_types_new_linked,
                                                    'evidence_linkage': (
                                                        'new_linked_pmids' if new_linked_pmids else
                                                        ('linked_pmids_but_no_new' if curr_pmids else 'no_linked_pmids')
                                                    ),
                                                    'label_trust': 'low' if result.get('version_history_incomplete') else 'high',
                                                    'validation_level': 'strict' if self.strict_condition_matching else 'lenient'
                                                }

                                                result['rcv_version_changes'].append(rcv_change)
                                                self.debug_stats['rcv_changes_found'] += 1

                                except Exception as e:
                                    logger.debug(f"Error extracting RCV diff for version {curr_version['version_id']}: {e}")

                        # VCV-level analysis
                        if self.training_cutoff_date:
                            curr_date = curr_version.get('date_parsed')
                            if curr_date and curr_date > self.training_cutoff_date:
                                self.debug_stats['temporal_filtered'] += 1
                                continue

                        prev_class = prev_version.get('classification', '').strip()
                        curr_class = curr_version.get('classification', '').strip()

                        if prev_class and curr_class and prev_class.lower() != curr_class.lower():
                            change_type = self.classify_change(prev_class, curr_class, var_id)

                            if change_type and change_type != 'not_relevant':
                                # Check for PMIDs
                                has_pmids_new = False
                                study_types_new = []
                                new_pmids = []
                                removed_pmids = []

                                if hasattr(self, 'version_xmls_cache'):
                                    prev_xml = self.version_xmls_cache.get(prev_version['version_id'])
                                    curr_xml = self.version_xmls_cache.get(curr_version['version_id'])

                                    if prev_xml and curr_xml:
                                        try:
                                            prev_pmids, _, prev_map = self._pmids_and_study_types_for_condition(
                                                ET.fromstring(prev_xml), '', prev_class
                                            )
                                            curr_pmids, _, curr_map = self._pmids_and_study_types_for_condition(
                                                ET.fromstring(curr_xml), '', curr_class
                                            )

                                            new_pmids = sorted(set(curr_pmids) - set(prev_pmids))
                                            removed_pmids = sorted(set(prev_pmids) - set(curr_pmids))
                                            has_pmids_new = len(new_pmids) > 0

                                            study_types_new = sorted({t for p in new_pmids
                                                                    for t in curr_map.get(p, [])} or ['unspecified'])

                                        except Exception as e:
                                            logger.debug(f"Error extracting PMID diff: {e}")

                                vcv_change = {
                                    'version_old': prev_version_num,
                                    'version_new': curr_version_num,
                                    'date_old': prev_version.get('date', ''),
                                    'classification_old': prev_class,
                                    'date_new': curr_version.get('date', ''),
                                    'classification_new': curr_class,
                                    'change_type': change_type,
                                    'source': 'VCV',
                                    'study_types_new': study_types_new,
                                    'has_citations': has_pmids_new,
                                    'evidence_linkage': 'new_pmids' if has_pmids_new else 'no_new_pmids',
                                    'new_pmids': new_pmids,
                                    'new_pmid_count': len(new_pmids),
                                    'removed_pmids': removed_pmids,
                                    'removed_pmid_count': len(removed_pmids),
                                    'label_trust': 'low' if result.get('version_history_incomplete') else 'high',
                                    'validation_level': 'strict' if self.strict_condition_matching else 'lenient'
                                }
                                result['vcv_version_changes'].append(vcv_change)
                                self.debug_stats['vcv_changes_found'] += 1

                return result

            except Exception as e:
                self.debug_stats['parse_errors'] += 1
                logger.error(f"Error processing variant {var_id}: {e}")
                return None

        # Process all filtered variants in parallel
        with ThreadPoolExecutor(max_workers=max_workers) as executor:
            future_to_id = {
                executor.submit(process_variant_with_versions, var_id, var_info): var_id
                for var_id, var_info in filtered_variants.items()
            }

            processed_count = 0
            failed_count = 0

            for future in as_completed(future_to_id):
                var_id = future_to_id[future]
                try:
                    result = future.result(timeout=180)
                    if result:
                        results.append(result)
                        processed_count += 1
                    else:
                        failed_count += 1

                    total_done = processed_count + failed_count
                    if total_done % 100 == 0:
                        elapsed = time.time() - start_time
                        if elapsed > 0:
                            print(f"[PROCESSING] Progress: {total_done}/{len(filtered_variants)} ({100*total_done/len(filtered_variants):.1f}%)")

                except Exception as e:
                    logger.error(f"Failed to process {var_id}: {e}")
                    failed_count += 1

        elapsed = time.time() - start_time
        print(f"\n[PROCESSING] Completed in {elapsed:.1f} seconds ({elapsed/60:.1f} minutes)")
        print(f"[PROCESSING] Successfully processed: {processed_count}")
        print(f"[PROCESSING] Failed: {failed_count}")

        return results

    def process_all_variants(self, gene_names: List[str], limit: int = None, max_workers: int = 10,
                             require_citations: bool = False, include_conflicts: bool = True,
                             fetch_versions: bool = True, unbiased_search: bool = False,
                             use_scv_reclassifications: bool = False,
                             skip_incomplete_histories: bool = True,
                             training_cutoff_date: str = None):
        """
        Main processing method - now using optimized two-pass approach
        """
        print("\n" + "="*80)
        print(f"PROCESSING VARIANTS IN GENES: {', '.join(gene_names)}")
        print("="*80)

        start_time = time.time()
        self.current_gene_names = gene_names

        # Set temporal cutoff if provided
        if training_cutoff_date:
            self.training_cutoff_date = self.parse_date(training_cutoff_date)
            print(f"[SETTINGS] Training cutoff date: {training_cutoff_date}")
            print(f"  → Only including data before this date for training")

        # Search for variants
        variation_ids = self.search_variants(gene_names, limit=limit, unbiased=unbiased_search)

        print(f"\n[PROCESS] Processing {len(variation_ids)} variants...")
        print(f"[SETTINGS] Unbiased search: {unbiased_search}")
        print(f"[SETTINGS] Fetch version history: {fetch_versions}")
        print(f"[SETTINGS] Include conflicts in output: {include_conflicts}")
        print(f"[SETTINGS] Use SCV reclassifications: {use_scv_reclassifications}")
        print(f"[SETTINGS] Use RCV level: {self.use_rcv_level}")
        print(f"[SETTINGS] Require citations: {require_citations}")
        print(f"[SETTINGS] Skip incomplete histories: {skip_incomplete_histories}")
        print(f"[SETTINGS] Strict condition matching: {self.strict_condition_matching}")
        print(f"[SETTINGS] Require PMID validation: {self.require_pmid_validation}")

        # Use OPTIMIZED processing
        all_results = self.process_variants_parallel_optimized(
            variation_ids,
            gene_names=gene_names,
            max_workers=max_workers,
            fetch_versions=fetch_versions,
            require_citations=require_citations,
            use_scv_reclassifications=use_scv_reclassifications
        )

        # Filter and organize results
        variants_with_reclassifications = []
        variants_with_conflicts = []
        variants_with_vcv_changes = []
        variants_with_rcv_changes = []

        for result in all_results:
            if not result:
                continue

            # Skip if incomplete history and flag is set
            if skip_incomplete_histories and result.get('version_history_incomplete'):
                self.debug_stats['skipped_incomplete_history'] += 1
                continue

            # Prepare variant record
            variant_record = {
                'variation_id': result['variation_id'],
                'vcv_accession': result.get('vcv_accession', ''),
                'rsid': result.get('rsid', ''),
                'genes': gene_names,
                'variant_name': result.get('name', '').strip(),
                'variant_type': result.get('variant_type', ''),
                'total_submissions': len(result.get('submissions', [])),
                'all_submissions': result.get('submissions', []),
                'version_history_incomplete': result.get('version_history_incomplete', False)
            }

            # Track SCV reclassifications (if enabled)
            if result.get('reclassifications'):
                variant_record['reclassifications'] = result['reclassifications']
                variants_with_reclassifications.append(variant_record.copy())

            # Track VCV version changes
            if result.get('vcv_version_changes'):
                variant_record['vcv_version_changes'] = result['vcv_version_changes']
                variants_with_vcv_changes.append(variant_record.copy())

            # Track RCV version changes
            if result.get('rcv_version_changes'):
                variant_record['rcv_version_changes'] = result['rcv_version_changes']
                variants_with_rcv_changes.append(variant_record.copy())

            # Track current conflicts
            if include_conflicts and result.get('current_conflicts') and result['current_conflicts'].get('has_vus_path_conflict'):
                variant_record['current_conflicts'] = result['current_conflicts']
                variants_with_conflicts.append(variant_record.copy())

        processing_time = time.time() - start_time
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        gene_str = '_'.join(gene_names)

        # FIXED: Use consistent counting
        total_processed_after_errors = len(all_results)

        # Prepare output with proper separation of concepts
        output_data = {
            'metadata': {
                'genes': gene_names,
                'extraction_date': datetime.now().isoformat(),
                'processing_time_seconds': processing_time,
                'processing_time_minutes': processing_time / 60,
                'total_variants_searched': len(variation_ids),
                'total_variants_fetched': self.debug_stats['variants_fetched_successfully'],
                'total_variants_processed': total_processed_after_errors,  # FIXED: Consistent
                'variants_pre_filtered': self.debug_stats['variants_pre_filtered'],
                'variants_with_meaningful_type': self.debug_stats['variants_with_meaningful_type_count'],
                'variants_with_reclassifications': len(variants_with_reclassifications),
                'variants_with_vcv_changes': len(variants_with_vcv_changes),
                'variants_with_rcv_changes': len(variants_with_rcv_changes),
                'variants_with_current_conflicts': len(variants_with_conflicts),
                'variants_with_incomplete_history': self.debug_stats.get('variants_with_incomplete_history', 0),
                'variants_skipped_incomplete': self.debug_stats.get('skipped_incomplete_history', 0),
                'temporal_filtered': self.debug_stats.get('temporal_filtered', 0),
                'condition_mismatches': self.debug_stats.get('condition_mismatches', 0),
                'empty_condition_mismatches': self.debug_stats.get('empty_condition_mismatches', 0),
                'splice_region_rejected': self.debug_stats.get('splice_region_rejected', 0),
                'deletions_rejected_no_size': self.debug_stats.get('deletions_rejected_no_size', 0),
                'variants_from_cache': self.debug_stats.get('variants_from_cache', 0),
                'versions_from_cache': self.debug_stats.get('versions_from_cache', 0),
                'scv_temporal_is_lower_bound': use_scv_reclassifications,
                'data_limitations': {
                    'scv_reclassifications': 'Disabled - not used in analysis' if not use_scv_reclassifications else 'Condition-aware with 30-day gap requirement',
                    'conflicts': 'Per-condition analysis' if include_conflicts else 'Not computed',
                    'vcv_changes': 'Aggregate level - may conflate phenotypes' if not self.use_rcv_level else 'Per-condition RCV analysis',
                    'rcv_changes': 'Per-condition with targeted PMID attribution' if self.use_rcv_level else 'Not computed',
                    'validation': 'Strict condition matching and PMID validation' if self.strict_condition_matching else 'Lenient matching'
                },
                'search_type': 'unbiased' if unbiased_search else 'filtered_conflicts',
                'processing_rate': total_processed_after_errors / processing_time if processing_time > 0 else 0,
                'training_cutoff_date': training_cutoff_date if training_cutoff_date else None,
                'optimization_savings': f"Saved {self.debug_stats['variants_pre_filtered']} variants from version fetching",
                'settings': {
                    'require_citations': require_citations,
                    'include_conflicts': include_conflicts,
                    'fetch_versions': fetch_versions,
                    'unbiased_search': unbiased_search,
                    'use_scv_reclassifications': use_scv_reclassifications,
                    'use_rcv_level': self.use_rcv_level,
                    'skip_incomplete_histories': skip_incomplete_histories,
                    'strict_condition_matching': self.strict_condition_matching,
                    'require_pmid_validation': self.require_pmid_validation
                },
                'debug_stats': dict(self.debug_stats)
            },
            'reclassifications': variants_with_reclassifications if use_scv_reclassifications else [],
            'vcv_version_changes': variants_with_vcv_changes,
            'rcv_version_changes': variants_with_rcv_changes,
            'current_conflicts': variants_with_conflicts if include_conflicts else []
        }

        output_file = f"{gene_str}_analysis_{timestamp}.json"
        with open(output_file, 'w', encoding='utf-8') as f:
            json.dump(output_data, f, indent=2, ensure_ascii=False, default=str)

        print(f"\n[SAVE] Results saved to: {output_file}")

        self.print_summary(output_data)

        return output_data

    def print_summary(self, data: Dict):
        """
        Print analysis summary with validation checks
        """
        print("\n" + "="*80)
        print("ANALYSIS COMPLETE")
        print("="*80)

        metadata = data['metadata']

        print(f"\n⚡ PERFORMANCE:")
        print(f"  Processing time: {metadata['processing_time_seconds']:.1f} seconds ({metadata['processing_time_minutes']:.1f} minutes)")
        print(f"  Variants searched: {metadata['total_variants_searched']}")
        print(f"  Variants fetched: {metadata['total_variants_fetched']}")
        print(f"  Variants pre-filtered: {metadata['variants_pre_filtered']}")
        print(f"  Variants processed: {metadata['total_variants_processed']}")
        print(f"  Processing rate: {metadata.get('processing_rate', 0):.1f} variants/second")
        print(f"  Variants from cache: {metadata.get('variants_from_cache', 0)}")
        print(f"  Versions from cache: {metadata.get('versions_from_cache', 0)}")

        print(f"\n📊 RESULTS SUMMARY:")
        print(f"  Search type: {metadata['search_type']}")
        print(f"  Variants with meaningful type: {metadata['variants_with_meaningful_type']}")
        print(f"  Variants with incomplete history: {metadata.get('variants_with_incomplete_history', 0)}")
        if metadata['settings'].get('skip_incomplete_histories'):
            print(f"  Variants skipped (incomplete): {metadata.get('variants_skipped_incomplete', 0)}")
        if metadata.get('training_cutoff_date'):
            print(f"  Temporal filtered (after cutoff): {metadata.get('temporal_filtered', 0)}")
        if metadata['settings'].get('strict_condition_matching'):
            print(f"  Condition mismatches detected: {metadata.get('condition_mismatches', 0)}")
            print(f"  Empty condition mismatches: {metadata.get('empty_condition_mismatches', 0)}")

        # Validation checks
        print(f"\n✅ VALIDATION CHECKS:")
        print(f"  Splice region variants rejected: {metadata.get('splice_region_rejected', 0)}")
        print(f"  Deletions rejected (no size): {metadata.get('deletions_rejected_no_size', 0)}")
        print(f"  Empty conditions not matched: {metadata.get('empty_condition_mismatches', 0)}")

        if metadata['settings']['use_scv_reclassifications']:
            print(f"\n  SCV TEMPORAL RECLASSIFICATIONS:")
            print(f"    Variants with SCV reclassifications: {metadata['variants_with_reclassifications']}")
            for change_type in ['upgraded_vus_to_pathogenic', 'upgraded_vus_to_likely_pathogenic',
                               'downgraded_pathogenic_to_vus', 'downgraded_likely_pathogenic_to_vus']:
                count = self.debug_stats.get(f'reclassifications_found_{change_type}', 0)
                if count > 0:
                    print(f"      {change_type}: {count}")

        if metadata['settings'].get('use_rcv_level'):
            print(f"\n  RCV VERSION CHANGES (PER-CONDITION - MOST ACCURATE):")
            print(f"    Variants with RCV changes: {metadata['variants_with_rcv_changes']}")
            print(f"    Total RCV changes found: {self.debug_stats.get('rcv_changes_found', 0)}")

            if data['rcv_version_changes']:
                # Analyze conditions and evidence
                conditions_seen = set()
                with_new_pmids = 0
                for var in data['rcv_version_changes']:
                    for change in var.get('rcv_version_changes', []):
                        conditions_seen.add(change.get('condition', 'unspecified'))
                        if change.get('new_linked_pmids'):
                            with_new_pmids += 1

                print(f"    Unique conditions tracked: {len(conditions_seen)}")
                print(f"    Changes with new PMIDs: {with_new_pmids}")

        print(f"\n  VCV VERSION CHANGES (AGGREGATE):")
        print(f"    Variants with VCV changes: {metadata['variants_with_vcv_changes']}")
        print(f"    Total VCV changes found: {self.debug_stats.get('vcv_changes_found', 0)}")

        print(f"\n  CURRENT CONFLICTS (labs disagree right now):")
        print(f"    Variants with VUS/Path conflicts: {metadata['variants_with_current_conflicts']}")

        # Calculate percentages
        if metadata['total_variants_processed'] > 0:
            total = metadata['total_variants_processed']
            print(f"\n  PERCENTAGES:")
            if metadata['settings']['use_scv_reclassifications']:
                print(f"    SCV Reclassifications: {100*metadata['variants_with_reclassifications']/total:.2f}%")
            if metadata['settings'].get('use_rcv_level'):
                print(f"    RCV Version Changes: {100*metadata['variants_with_rcv_changes']/total:.2f}%")
            print(f"    VCV Version Changes: {100*metadata['variants_with_vcv_changes']/total:.2f}%")
            print(f"    Current conflicts: {100*metadata['variants_with_current_conflicts']/total:.2f}%")

        # Debug stats
        print(f"\n🔍 DEBUG STATS:")
        print(f"  Fetch success rate: {100*metadata['total_variants_fetched']/metadata['total_variants_searched']:.1f}%")
        print(f"  Valid XML: {self.debug_stats['variants_with_valid_xml']}")
        print(f"  Invalid XML: {self.debug_stats['variants_with_invalid_xml']}")
        print(f"  Current conflicts found: {self.debug_stats.get('current_conflicts_found', 0)}")
        print(f"  Versions fetched: {self.debug_stats.get('versions_fetched', 0)}")


def main():
    # Configuration
    EMAIL = "yl8889@nyu.edu"
    API_KEY = "92c067f1875f6abf0dfbf5d0d57758beff09"

    # Gene names to search
    GENE_NAMES = ["BRCA1"]

    # Processing options
    LIMIT_VARIANTS = None  # Set to None for all variants
    MAX_WORKERS = 15

    # RECOMMENDED SETTINGS FOR VALID TRAINING DATA
    UNBIASED_SEARCH = True  # Get ALL variants (not just conflicts)
    REQUIRE_CITATIONS = False  # Don't gate changes, just annotate evidence
    INCLUDE_CONFLICTS = True  # Include but compute per-condition
    FETCH_VERSIONS = True  # Required for version-based analysis
    USE_SCV_RECLASSIFICATIONS = True  # Enable for more data
    USE_RCV_LEVEL = True  # Enable per-condition RCV analysis (MOST ACCURATE)
    SKIP_INCOMPLETE_HISTORIES = True  # IMPORTANT: Skip incomplete histories for training

    # Temporal split for training/testing
    TRAINING_CUTOFF_DATE = None  # Set to e.g., "2024-01-01" for temporal split

    print("\n🚀 FIXED VERSION WITH DEBUGGING")
    print("  ✅ Condition extraction fixed (uses MedGen IDs as fallback)")
    print("  ✅ Classification normalization preserves Likely vs non-Likely")
    print("  ✅ PMIDs deduplicated")
    print("  ✅ Extensive debug logging for RCV/SCV/conditions")
    print("  ✅ Cache reuse from previous run")
    print("\n📋 Settings:")
    print(f"  Genes: {', '.join(GENE_NAMES)}")
    print(f"  Variants to process: {LIMIT_VARIANTS if LIMIT_VARIANTS else 'ALL'}")
    print(f"  Search type: {'UNBIASED (all variants)' if UNBIASED_SEARCH else 'FILTERED (conflicts only)'}")
    print(f"  Fetch version history: {FETCH_VERSIONS}")
    print(f"  Use SCV reclassifications: {USE_SCV_RECLASSIFICATIONS}")
    print(f"  Use RCV level analysis: {USE_RCV_LEVEL}")
    print(f"  Require citations: {REQUIRE_CITATIONS}")
    print(f"  Skip incomplete histories: {SKIP_INCOMPLETE_HISTORIES}")
    print(f"  Training cutoff date: {TRAINING_CUTOFF_DATE if TRAINING_CUTOFF_DATE else 'None (use all data)'}")
    print("="*80)

    api = ClinVarParallelAPI(email=EMAIL, api_key=API_KEY)
    api.use_rcv_level = USE_RCV_LEVEL  # Set the RCV flag
    api.strict_condition_matching = True  # Use strict condition matching for accuracy
    api.require_pmid_validation = True  # Require PMID validation

    try:
        result_data = api.process_all_variants(
            gene_names=GENE_NAMES,
            limit=LIMIT_VARIANTS,
            max_workers=MAX_WORKERS,
            require_citations=REQUIRE_CITATIONS,
            include_conflicts=INCLUDE_CONFLICTS,
            fetch_versions=FETCH_VERSIONS,
            unbiased_search=UNBIASED_SEARCH,
            use_scv_reclassifications=USE_SCV_RECLASSIFICATIONS,
            skip_incomplete_histories=SKIP_INCOMPLETE_HISTORIES,
            training_cutoff_date=TRAINING_CUTOFF_DATE
        )

        print("\n✅ Processing complete!")

        # Print key fixes
        print("\n🎯 FIXES APPLIED:")
        print("  • Condition extraction uses MedGen IDs when names are missing")
        print("  • Classification categories preserve Likely pathogenic vs Pathogenic")
        print("  • PMIDs are deduplicated before counting")
        print("  • Debug logging shows exactly where RCV/SCV extraction fails")
        print("  • Cache is reused from previous run for fast execution")

        if USE_RCV_LEVEL:
            print("\n📌 CHECK THE OUTPUT:")
            print("   Look for RCV changes - should be > 0 now with fixes")
            print("   Check debug output for [DEBUG-RCV], [DEBUG-SCV], [DEBUG-CONDITION] messages")

    except Exception as e:
        print(f"\n❌ Error during processing: {e}")
        import traceback
        traceback.print_exc()


if __name__ == "__main__":
    main()
```

```python
def extract_rcv_data_correctly(xml_file):
    """
    Correctly extract RCV data with fixed paths
    """
    tree = ET.parse(xml_file)
    root = tree.getroot()

    print(f"Extracting from: {xml_file}\n")

    # Get all RCV elements
    rcv_elements = root.findall('.//ReferenceClinVarAssertion')
    print(f"Found {len(rcv_elements)} RCV elements\n")

    rcv_data = []

    for rcv in rcv_elements:
        # Get RCV ID - FIXED PATH
        rcv_acc_elem = rcv.find('./ClinVarAccession[@Type="RCV"]')
        if rcv_acc_elem is not None:
            rcv_id = rcv_acc_elem.get('Acc')  # 'Acc' not 'Accession'!
            version = rcv_acc_elem.get('Version')

            # Get condition/trait
            condition_parts = []
            for trait in rcv.findall('.//TraitSet/Trait'):
                name_elem = trait.find('./Name/ElementValue[@Type="Preferred"]')
                if name_elem is not None and name_elem.text:
                    condition_parts.append(name_elem.text.strip())

            # Get classification - try multiple paths
            classification = None
            date_eval = None

            # Path 1: Under Classifications
            classifications = rcv.find('./Classifications')
            if classifications is not None:
                # Try GermlineClassification first
                germ_class = classifications.find('./GermlineClassification')
                if germ_class is not None:
                    desc = germ_class.find('./Description')
                    if desc is not None:
                        classification = desc.text
                    date_eval = germ_class.get('DateLastEvaluated')

            # Path 2: Under ClinicalSignificance (older format)
            if not classification:
                clin_sig = rcv.find('./ClinicalSignificance')
                if clin_sig is not None:
                    desc = clin_sig.find('./Description')
                    if desc is not None:
                        classification = desc.text
                    if not date_eval:
                        date_eval = clin_sig.get('DateLastEvaluated')

            rcv_entry = {
                'rcv_id': f"{rcv_id}.{version}" if rcv_id and version else rcv_id,
                'rcv_base': rcv_id,
                'version': version,
                'condition': ' | '.join(condition_parts) if condition_parts else 'Unknown',
                'classification': classification or 'Not found',
                'date_evaluated': date_eval or 'No date',
                'rcv_internal_id': rcv.get('ID')
            }

            rcv_data.append(rcv_entry)

            print(f"RCV: {rcv_entry['rcv_id']}")
            print(f"  Condition: {rcv_entry['condition']}")
            print(f"  Classification: {rcv_entry['classification']}")
            print(f"  Date: {rcv_entry['date_evaluated']}")
            print()

    return rcv_data

for variant_id in ["1065961", "927378", "867266"]:
    file_name = f"test_clinvarset_{variant_id}.xml"
    if Path(file_name).exists():
        print(f"\n{'='*60}")
        print(f"Variant {variant_id}")
        print('='*60)
        rcv_data = extract_rcv_data_correctly(file_name)

        # Check if we have per-condition data
        if len(rcv_data) > 1:
            print(f"✅ SUCCESS: Found {len(rcv_data)} condition-specific RCV records")
            print("This is what we need for ML training!")
```
